### Wordnet Lemmetizar
Lemmatization technique is like stemming. The output we will get after lemmatization is called "lemma". which is a root word rather than root stem, the output of stemming. after lemmatization we will be getting a valid word that means the same thing

NLTK provides WordNetLemmatize class which is a thin wrapper around the wordnet corpus. this class uses morphy() function to the WordNet CorpusReader class to find a lemma. let us understand it wit an example -

```python

#pos - Noun - 'n'
#pos - verb - 'v'
#pos - adjective - 'a'
#pos - adverb - 'r'

from nltk.stem import WordNetLemmatizer
Lemmatizer = WordNetLemmatizer()
Lemmatizer.lemmatize('going',pos = 'v')
```

```python
words = ["eating", "eats", "eaten", "writing","writes", "programming","programs","history","finally","finalized"]
```

```python
for word in words:
    print(word+"----->"+lemmatizer.lemmatize(word))
```
