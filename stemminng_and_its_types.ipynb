### RegexpStemmer Class

NLTK has RegexpStemmer class with the help of which we can easily implement Regular Expression Stemmer Algrorithms. It basically takes a single regular expression and remove any prefix or suffix that matches the expression. Let us see an example

```python
from nltk import RegexpStemmer
```

```python
reg_stemmer = RegexpStemmer('ing$|s$|e$|ed$', min=4)
```

regexpstemmer having $ sign mean ing,e,s,able will be removed, with the minimum length of 4.

```python
reg_stemmer.stem("eating")
```

```python
reg_stemmer.stem("rolling")
```
